```python
import matplotlib.pyplot as plt

years = [ 1992 , 1994, 1996 , 1998 , 2000 , 2002 , 2004 , 2006 , 2008 , 2010 ]
runs = [ 500 , 700 , 1100 , 1000 , 900 , 1300 , 1200 ,1400 , 850 , 1250 ]

plt.xlabel("Year")
plt.ylabel("Run's Scored")
plt.title("Sachin,s Sir Runs Over Years")
plt.plot(years,runs)
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt
years = [2002,2004,2006,2008,2010,2012,2014,2016,2018]
Virat = [0,0,300,450,600,900,650,1100,1250]
Sehwag = [670,900,450,800,1150,850,600,0,0]
Sachin = [1100,980,990,1200,760,1050,880,0,0]

plt.plot(years,Virat,label="virat kohli")
plt.plot(years,Sehwag,label="viru paji")
plt.plot(years,Sachin,label="Sachin sir")

plt.xlabel("Runs Scored")
plt.ylabel("Years")
plt.title("viru sir vs kohli vs Sachin sir Runs")
plt.legend()
plt.show()
```

```python
with plt.xkcd():

    plt.plot(years,Sehwag,label="viru paji")
    plt.plot(years,Virat,label="virat kohli")
    plt.plot(years,Sachin,label="Sachin sir")

    plt.xlabel("Runs Scored")
    plt.ylabel("Years")
    plt.title("viru sir vs kohli Runs")
    plt.grid(True)
    plt.legend()
    plt.show()
```
