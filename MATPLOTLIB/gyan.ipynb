```python
import matplotlib.pyplot as plt
years = [2002,2004,2006,2008,2010,2012,2014,2016,2018]
Virat = [0,0,300,450,600,900,650,1100,1250]
Sehwag = [670,900,450,800,1150,850,600,0,0]
Sachin = [1100,980,990,1200,760,1050,880,0,0]

plt.plot(years,Virat,label="virat kohli")
plt.plot(years,Sehwag,label="viru paji")
plt.plot(years,Sachin,label="Sachin sir")

plt.xlabel("Runs Scored")
plt.ylabel("Years")
plt.title("viru sir vs kohli vs Sachin sir Runs")
plt.legend()
plt.show()
```

```python
with plt.xkcd():

    plt.plot(years,Sehwag,label="viru paji")
    plt.plot(years,Virat,label="virat kohli")
    plt.plot(years,Sachin,label="Sachin sir")

    plt.xlabel("Runs Scored")
    plt.ylabel("Years")
    plt.title("viru sir vs kohli Runs")
    plt.legend()
    plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

Day = ["mon","tue","wed","thu","fri"]
Hours = [ 5 , 4 , 6 , 7 , 5 ]
highest = max(Hours)
index = Hours.index(highest)

# Maximum attendance wale din ki values
max_day = Day[index]
max_hours = Hours[index]

plt.plot( Day , Hours, linestyle="--", marker="o")
plt.plot(max_day, max_hours, color="red", marker="o", linestyle="None", label="Highest Attendance")
plt.xlabel("Days of week")
plt.ylabel("Hours of Study on that day")
plt.title("study hrs in a day of week")
plt.legend()
plt.grid(True)
plt.show()
```
